# Exploratory Data Analysis (EDA)
This notebook explores the training data for the Text Mining project.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
import os
```

```python
# Load the training data
data_path = os.path.join('..', 'data', 'train.csv')
df = pd.read_csv(data_path)
df.head()
```

## Class Distribution
Visualize the distribution of sentiment labels.

```python
# Plot class distribution
label_map = {0: 'Bearish', 1: 'Bullish', 2: 'Neutral'}
df['label_name'] = df['label'].map(label_map)
sns.countplot(x='label_name', data=df, palette='Set2')
plt.title('Class Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.show()
```

## Tweet Length Analysis
Explore the distribution of tweet lengths.

```python
# Add tweet length column
df['tweet_length'] = df['text'].str.len()
sns.histplot(df['tweet_length'], bins=30, kde=True)
plt.title('Tweet Length Distribution')
plt.xlabel('Number of Characters')
plt.ylabel('Frequency')
plt.show()
```

## Word Cloud per Sentiment
Visualize the most common words for each sentiment.

```python
# Generate word clouds for each sentiment
for label, name in label_map.items():
    text = ' '.join(df[df['label'] == label]['text'])
    wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text)
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.title(f'Word Cloud for {name} Tweets')
    plt.show()
```
